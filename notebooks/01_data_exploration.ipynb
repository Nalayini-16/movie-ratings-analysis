```python
def convert_votes(value):
    value = value.strip('()')
    if 'M' in value:
        return float(value.replace('M', ''))
    elif 'K' in value:
        return float(value.replace('K', ''))
    else:
        return float(value)
df['numberof_ratings'] = df['numberof_ratings'].apply(convert_votes)
```

```python
def convert_duration(value):
    parts = value.split('h')
    hours = int(parts[0].strip())
    minutes = int(parts[1].replace('m', '').strip())
    return hours * 60 + minutes
df['duration'] = df['duration'].apply(convert_duration)
```

```python
def convert_duration(value):
    hours = 0
    minutes = 0
    if 'h' in value:
        parts = value.split('h')
        hours = int(parts[0].strip())
        remainder = parts[1].replace('m', '').strip()
        if remainder:  # only convert if not empty
            minutes = int(remainder)
    return hours * 60 + minutes

df['duration'] = df['duration'].apply(convert_duration)
```
